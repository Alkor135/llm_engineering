```python
"""
Для загрузки модели SentenceTransformer и сохранения её в кэше с использованием токена Hugging Face.
"""

from sentence_transformers import SentenceTransformer
import os
from dotenv import load_dotenv

# Указываем директорию для кэша
cache_dir = "C:\\Users\\Alkor\\model_cache"

# Загрузка переменных окружения из .env файла
load_dotenv(override=True)
hf_api_token = os.getenv('HF_API_TOKEN')  # Получаем токен из переменной окружения

# Проверяем, получен ли токен
if not hf_api_token:
    raise ValueError("Токен HF_API_TOKEN не найден в переменных окружения. Проверьте файл .env.")

# Загружаем модель и сохраняем её в кэш
model = SentenceTransformer(
    'all-MiniLM-L6-v2',
    cache_folder=cache_dir,
    token=hf_api_token  # Передаем реальный токен
)
model.save(cache_dir)  # Сохраняем модель в указанную директорию
print(f"Модель сохранена в {cache_dir}")
```
